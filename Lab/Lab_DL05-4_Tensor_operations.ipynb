```python
# Access
arr_t = torch.Tensor([[1,2],[2,3]])
arr_t[0,1]
```

```python
#
# Slicing
#
```

```python
t = torch.Tensor([[1,2,3,4], [2,3,4,5], [5,6,7,8]])
print(t)
```

```python
t[:2]
```

```python
t[:,1:]
```

```python
t[1:,1:3]
```

```python
t[1:, 1:3] = 0
print(t)
```

```python
#
# Negative slicing
#
```

```python
t = torch.Tensor([[1,2,3,4], [2,3,4,5], [5,6,7,8]])
print(t)
```

```python
t[:,:-1]
```

```python
t[:,-3:-1]
```

```python
#
# Shape & Transpose (matrix)
#
```

```python
X = torch.Tensor([[1,2,3], [4,5,6]])
```

```python
X
```

```python
X.T
```

```python
print(X.shape)
print(X.T.shape)
```

```python
#
# Sum
#
```

```python
print(X.sum())
```

```python
print(X.sum(0))
print(X.sum(1))
```

```python
print(X.sum(0, keepdim = True))
print(X.sum(1, keepdim = True))
```

```python
#
# Mean
#
```

```python
print(X.mean())
print(X.mean(0))
print(X.mean(1))
```

```python
#
# Binary Operators
#
```

```python
X = torch.Tensor([[1,2,3], [4,5,6]])
Y = torch.Tensor([[1,0,2], [1,0,1]])
```

```python
X
```

```python
Y
```

```python
X+Y
```

```python
X*Y
```

```python
torch.matmul(X.T, Y)
```

```python
torch.matmul(X, Y.T)
```
