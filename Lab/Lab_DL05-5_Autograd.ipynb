```python
# torch.autograd is PyTorch's autpmatic differsntiation engine that powers neural network training
# Initialization: Set requires_grad to True if you want to track the gradient
import torch

w = torch.randn(2, requires_grad = True)
x = torch.ones(2)

print("w", w.requires_grad)
print("x", x.requires_grad)
```

```python
# 1. Initialize
w = torch.randn(2, requires_grad = True)
x = torch.Tensor([1,2])

# 2. Predict output
y_hat = torch.inner(w,x)

# 2-2. Intermediate results
print(x)
print(w)
print(y_hat)
print('\n')

# 3. Compute loss
loss = (x.mean()-y_hat)**2
print(loss)

# 4. Backpropagation
loss.backward()

# 4-2. Accessing the gradient
w.grad
```

```python
# Manual parameter update
lr = 0.1

with torch.no_grad():
    w = w-lr * w.grad
print(w.requires_grad)
w.requires_grad = True
```
