```python
# Data preparation & Import
import numpy as np
import torch

# XOR data(numpy)
x_seeds = np.array([(0,0), (1,0), (0,1), (1,1)], dtype = np.float32)
y_seeds = np.array([0,1,1,0])

N = 1000
idxs = np.random.randint(0,4,N)

X = x_seeds[idxs]
Y = y_seeds[idxs]

X += np.random.normal(scale = 0.25, size = X.shape)
```

```python
# Shallow NN: Model
from torch import nn

class shallow_neural_network(nn.Module):
    def __init__(self, num_input_features, num_hiddens):
        super().__init__()
        self.num_input_features = num_input_features
        self.num_hiddens = num_hiddens
        
        self.linear1 = nn.Linear(num_input_features, num_hiddens)
        self.linear2 = nn.Linear(num_hiddens, 1)
        
        self.tanh = torch.nn.Tanh()
        self.sigmoid = torch.nn.Sigmoid()
        
    def forward(self, x):
        z1 = self.linear1(x)
        a1 = self.tanh(z1)
        z2 = self.linear2(a1)
        a2 = self.sigmoid(z2)
        return a2
```

```python
# Shallow NN: Training
# Version1: Stochastic Gradient Descent
# Foreach epoch:
    # Foreach training example
        # - Compute loss
        # - Update model
from torch import optim

num_epochs = 100
lr = 1.0
num_hiddens = 3

model = shallow_neural_network(2, num_hiddens)
optimizer = optim.SGD(model.parameters(), lr = lr)
loss = nn.BCELoss()

for epoch in range(num_epochs):
    cost = 0.0
    for x, y in zip(X, Y):
        x_torch = torch.from_numpy(x)
        y_torch = torch.FloatTensor([y])
        
        y_hat = model(x_torch)
        
        loss_fn = loss(y_hat, y_torch)
        optimizer.zero_grad()
        loss_fn.backward()
        optimizer.step()
        
        cost += loss_fn.item()
    
    cost = cost / len(X)
    if epoch % 10 == 0:
        print(epoch, cost)
```

```python
# Shallow NN: Training
# Version2: Gradient Descent
# Foreach epoch:
    # - Compute cost
    # - Update model
num_epochs = 100
lr = 1.0
num_hiddens = 3

model = shallow_neural_network(2, num_hiddens)
optimizer = optim.SGD(model.parameters(), lr = lr)
loss = nn.BCELoss()

for epoch in range(num_epochs):
    optimizer.zero_grad()
    
    cost = 0.0
    for x,y in zip(X,Y):
        x_torch = torch.from_numpy(x)
        y_torch = torch.FloatTensor([y])
        
        y_hat = model(x_torch)
        
        loss_val = loss(y_hat, y_torch)
        cost += loss_val;
    
    cost = cost / len(X)
    cost.backward()
    optimizer.step()
    
    if epoch % 10 == 0:
        print(epoch, cost)
```

```python
# Test
for x, y in zip(x_seeds, y_seeds):
    print(x)
    x_torch = torch.FloatTensor(x)
    y_hat = model(x_torch)
    print(y, y_hat.item())
```
